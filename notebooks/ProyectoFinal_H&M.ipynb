```python
def explorar_datos(datos, nombre):
    """
    Muestra información básica de un DataFrame.

    Parámetros
    ----------
    datos : pd.DataFrame
        Tabla que queremos explorar.
    nombre : str
        Nombre de la tabla.
    """

    print(f"\n{'=' * 45}")
    print(f"EXPLORACIÓN: {nombre.upper()}")
    print(f"{'=' * 45}")

    print(f"\nNúmero de filas: {datos.shape[0]:,}")
    print(f"Número de columnas: {datos.shape[1]}")

    print("\nTIPOS DE DATOS:")
    print(datos.dtypes)

    print("\nVALORES NULOS:")
    nulos = pd.DataFrame({
        "Nulos": datos.isna().sum(),
        "Porcentaje (%)": (
            datos.isna().mean() * 100
        ).round(2)
    })

    display(nulos)

    print("\nFILAS DUPLICADAS:")
    print(datos.duplicated().sum())

    print("\nPRIMERAS FILAS:")
    display(datos.head())
```

### 4.4. Validación global del dataset procesado

Tras comprobar la estructura y calidad de una muestra inicial, se realiza una validación del archivo completo de transacciones filtradas.

El objetivo es verificar que todos los registros pertenecen al periodo seleccionado, comprobar la ausencia de valores nulos en las variables principales y detectar posibles precios no válidos.

Debido al volumen de datos, la validación se realiza mediante procesamiento por bloques (*chunks*), optimizando el uso de memoria.

```python

# 4.4. Validación global del dataset procesado

total_filas = 0
fechas_minimas = []
fechas_maximas = []
total_nulos = 0
precios_no_validos = 0
filas_fuera_periodo = 0

for bloque in pd.read_csv(
    ruta_transacciones_preparadas,
    chunksize=500_000,
    dtype={"article_id": "string"},
    parse_dates=["t_dat"]
):
    total_filas += len(bloque)

    fechas_minimas.append(bloque["t_dat"].min())
    fechas_maximas.append(bloque["t_dat"].max())

    total_nulos += bloque.isnull().sum().sum()

    precios_no_validos += (bloque["price"] <= 0).sum()

    filas_fuera_periodo += (
        ~bloque["t_dat"].between(
            "2019-09-23", "2020-09-22"
        )
    ).sum()

print("Total de transacciones:", f"{total_filas:,}")
print("Fecha mínima:", min(fechas_minimas))
print("Fecha máxima:", max(fechas_maximas))
print("Valores nulos:", total_nulos)
print("Precios no válidos:", precios_no_validos)
print("Registros fuera del periodo:", filas_fuera_periodo)
```

**Resultados de la validación global**

Se ha validado el archivo de transacciones correspondiente al periodo comprendido entre el 23 de septiembre de 2019 y el 22 de septiembre de 2020, con un total de **14.898.423 registros**.

Los resultados confirman que:

- Todas las transacciones se encuentran dentro del periodo seleccionado.
- No se han detectado valores nulos.
- No existen precios iguales o inferiores a cero.
- Las fechas mínima y máxima coinciden con los límites temporales establecidos.

Por tanto, el dataset de transacciones supera las comprobaciones iniciales de integridad y consistencia, y puede utilizarse para continuar con las siguientes etapas de preparación y análisis.

No ha sido necesario eliminar registros durante esta fase de validación.

### 8.2. Cálculo de indicadores comerciales generales (KPI)

Se calculan los principales indicadores de actividad comercial de H&M durante el periodo comprendido entre el 23 de septiembre de 2019 y el 22 de septiembre de 2020.

Los indicadores seleccionados son el número de unidades registradas, el número de clientes compradores únicos y la cantidad de referencias diferentes vendidas.

Estos KPI proporcionan una visión general del comportamiento comercial y servirán como punto de partida para los análisis posteriores por categorías, segmentos de clientes y canales de venta.

Debido a la ausencia de identificadores de pedido, cada registro se considera una unidad de artículo, sin asumir que equivale a un pedido independiente.

```python

# 8.2. KPI generales de H&M

kpi_generales = con.execute("""
    SELECT
        COUNT(*) AS unidades_registradas,
        COUNT(DISTINCT customer_id) AS clientes_compradores,
        COUNT(DISTINCT article_id) AS articulos_vendidos
    FROM transacciones
""").df()

display(kpi_generales)
```

### 8.3. Evolución temporal de la demanda

Se analiza la evolución mensual del volumen de artículos vendidos y del número de clientes compradores únicos durante el periodo seleccionado.

El objetivo es identificar variaciones en la actividad comercial, posibles patrones estacionales y meses con mayor o menor demanda.

El análisis permitirá contextualizar posteriormente el rendimiento de las categorías de productos y los canales de venta.

```python

# 8.3. Evolución mensual de la demanda

ventas_mensuales = con.execute("""
    SELECT
        anio_mes,
        COUNT(*) AS unidades_vendidas,
        COUNT(DISTINCT customer_id) AS clientes_unicos,
        COUNT(DISTINCT article_id) AS articulos_diferentes
    FROM transacciones
    GROUP BY anio_mes
    ORDER BY anio_mes
""").df()

display(ventas_mensuales)
```
